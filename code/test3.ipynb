```python
from cubeio import *
pre1 = cubeio.load('4238_4','predicted' )
pre2 = cubeio.load('4349_5', 'predicted')
lat1 = pre1.lat
lon1 = pre1.lon
reg1 = pre1.reg
lat2 = pre2.lat
lon2 = pre2.lon
reg2 = pre2.reg
lon = np.hstack([lon1, lon2])
lat = np.hstack([lat1, lat2])
reg = np.hstack([reg1, reg2])
```

```python
import matplotlib.pyplot as plt
reg[reg<=0]=np.nan
plt.scatter(lon, lat, c=reg, cmap='viridis', s=3)
plt.colorbar()  # 添加颜色条
plt.xlabel('X')
plt.ylabel('Y')
plt.title('pcolormesh示例')
plt.show() 

input("Press Enter to continue...")
```

```python
import glob
result = []
# 匹配所有以 _processed.pkl 结尾的文件
for name in glob.glob(os.path.join(config.py_path , 'predicted' , '*_predicted.pkl')):
    filename = os.path.basename(name)
    cube_name = filename[:-len('_processed.pkl')]
    result.append(cube_name)
print(result)
lats = np.zeros(0)
lons = np.zeros(0)
regs = np.zeros(0)
for name in result:
    cube = cubeio.load(name, 'predicted')
    lats = np.hstack([lats, cube.lat])
    lons = np.hstack([lons, cube.lon])
    regs = np.hstack([regs, cube.reg])
plt.scatter(lons, lats, c=regs, cmap='viridis', s=3)
plt.colorbar()  # 添加颜色条
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Points!!')
plt.show() 
input("Press Enter to continue...")
```
